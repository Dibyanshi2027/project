```python
import pandas as pd

df = pd.read_excel(r"C:\Startup-Investment-Analysis\data\Shark_Tank_India.xlsx.xlsx")

print(df.head())
```

```python
df.info()
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()

print("Duplicates removed successfully")
```

```python
print("Total Startups:", len(df))
```

```python
print("Total Investment:", df['Amount Invested lakhs'].sum())
```

```python
industry = df['Industry'].value_counts()

print(industry)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

industry.plot(kind='bar')

plt.title('Top Industries')

plt.xlabel('Industry')

plt.ylabel('Count')

plt.show()
```

```python
investors = df['Partners'].value_counts()

print(investors)
```

```python
plt.figure(figsize=(10,6))

investors.plot(kind='bar')

plt.title('Top Investors')

plt.xlabel('Investor')

plt.ylabel('Number of Investments')

plt.show()
```

```python
top_funding = df.sort_values(
    by='Amount Invested lakhs',
    ascending=False
)

print(
    top_funding[[
        'Startup Name',
        'Industry',
        'Amount Invested lakhs'
    ]].head(10)
)
```

```python
import seaborn as sns

plt.figure(figsize=(10,6))

sns.histplot(df['Equity Taken %'], bins=20)

plt.title('Equity Distribution')

plt.xlabel('Equity %')

plt.show()
```

```python
industry_investment = df.groupby('Industry')[
    'Amount Invested lakhs'
].sum().sort_values(ascending=False)

print(industry_investment)
```

```python
plt.figure(figsize=(12,6))

industry_investment.plot(kind='bar')

plt.title('Investment by Industry')

plt.xlabel('Industry')

plt.ylabel('Investment Amount')

plt.show()
```

```python
df.to_csv(
    r"C:\Startup-Investment-Analysis\data\cleaned_shark_tank.csv",
    index=False
)

print("Cleaned dataset saved successfully")
```
